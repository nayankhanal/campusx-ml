```python
df = pd.DataFrame()
```

```python
df
```

```python
for i in range(1,132):
  response = requests.get(f"https://api.themoviedb.org/3/tv/top_rated?api_key=957c3f6ceb4f114536342ddf1a867223&language=en-US&page={i}")
  results = response.json()['results']
  temp_df = pd.DataFrame(results)[['id','name','overview','first_air_date','popularity','vote_average','vote_count']]
  df = pd.concat([df, temp_df], ignore_index=True)
```
